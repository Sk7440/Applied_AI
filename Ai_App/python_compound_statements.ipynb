# Python Compound Statements

A compound statement consists of one or more "clauses"; where the clause begins with a keyword and is followed by a colon and an indented block of code. This notebook demonstrates the most common compound statements in Python.

## 1. The `if` Statement

Used for conditional execution.

```python
x = 10
if x > 15:
    print("x is greater than 15")
elif x > 5:
    print("x is greater than 5 but less than or equal to 15")
else:
    print("x is 5 or less")
```

## 2. The `for` Statement

Used for iterating over a sequence (list, tuple, dictionary, set, or string).

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}s")

print("\nCounting with range:")
for i in range(3):
    print(i)
```

## 3. The `while` Statement

Used to repeat a block of code as long as a condition is true.

```python
count = 0
while count < 3:
    print(f"Count is {count}")
    count += 1
```

## 4. The `match` Statement (Structural Pattern Matching)

Introduced in Python 3.10, similar to switch/case in other languages.

```python
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 404:
            return "Not found"
        case 500 | 503:
            return "Server error"
        case _:
            return "Something else"

print(http_error(404))
print(http_error(500))
print(http_error(200))
```

## 5. The `try` Statement (Exception Handling)

Used to handle errors gracefully.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
else:
    print("Division successful!")
finally:
    print("Execution of try-except block finished.")
```

## 6. The `with` Statement (Context Managers)

Used to ensure resources are properly managed (e.g., closing a file).

```python
with open("example.txt", "w") as f:
    f.write("Hello, this is a test file.")

with open("example.txt", "r") as f:
    print(f.read())
```

## 7. The `def` Statement (Function Definitions)

Used to define reusable blocks of code.

```python
def greet(name, greeting="Hello"):
    """Greets a person with a custom greeting."""
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", "Welcome"))
```

## 8. The `class` Statement (Class Definitions)

Used to create user-defined objects and classes.

```python
class Dog:
    species = "Canine"

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        return f"{self.name} says Woof!"

my_dog = Dog("Buddy", "Golden Retriever")
print(f"{my_dog.name} is a {my_dog.breed} ({my_dog.species}).")
print(my_dog.bark())
```
