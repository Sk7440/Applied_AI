```python
import os
import requests
import base64
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")
if not api_key:
    raise ValueError("GEMINI_API_KEY is not set in the .env file.")

# Gemini Developer API Imagen endpoint
url = f"https://generativelanguage.googleapis.com/v1beta/models/imagen-3.0-generate-002:predict?key={api_key}"

headers = {
    "Content-Type": "application/json"
}

payload = {
    "instances": [
        {
            "prompt": "A gourmet nano banana dish plated elegantly in a high-end restaurant, futuristic glowing Gemini aesthetic, professional food photography, 8k"
        }
    ],
    "parameters": {
        "sampleCount": 1,
        "aspectRatio": "1:1",
        "outputOptions": {
            "mimeType": "image/png"
        }
    }
}

print("Requesting image from Google Developer API...")
response = requests.post(url, headers=headers, json=payload)

if response.status_code == 200:
    data = response.json()
    predictions = data.get("predictions", [])
    if predictions:
        # Decode the returned base64 bytes
        b64_image = predictions[0]["bytesBase64Encoded"]
        image_bytes = base64.b64decode(b64_image)
        
        output_file = "generated_image.png"
        with open(output_file, "wb") as f:
            f.write(image_bytes)
        print(f"Success! Image saved to {output_file}")
    else:
        print("No image data found in response:", data)
else:
    print(f"Error ({response.status_code}): {response.text}")
```
